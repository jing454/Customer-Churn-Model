```python
import matplotlib.pyplot as plt

import pandas as pd
import numpy as np

df = pd.read_csv("../Cleaned_Data/Cleaned_Churn.csv")

df.head()
```

```python
#checking value counts of churn column
df["churn"].value_counts()
```

```python
#storing into variable for easier graphing

churn_true = 0 
churn_false = 0

churn_true = (df["churn"] == "Yes").sum()
print(churn_true)

churn_false = (df["churn"] == "No").sum()
print(churn_false)
```

```python
#storing results in array for plot
churn_result = np.array([churn_true, churn_false])
churn_label = np.array(["churn", "Didn't churn"])

#plotting
plt.bar(churn_label, churn_result)
plt.ylabel("Customers")
plt.title("Total Churn VS No Churn Results")
plt.grid(axis = "y")
plt.show()
```

```python
#Calculating Churn Percentage

churn_rate = (churn_true / (churn_true + churn_false)) * 100
print(f'The churn rate is {round(churn_rate, 2)}% (rounded to nearest 2 decimal places)')
```

```python
#Let's now graph all of them into one
x = np.arange(3)

churn_yes = np.array([5157,1044,642])
churn_no = np.array([6785,3946,2426])
width = 0.4

plt.bar(x-width/2, churn_yes, width, color="red", label="Didn't Churn")
plt.bar(x+width/2, churn_no, width, color="blue", label="Churn")

plt.xticks(x, ["Month-Month", "One year", "Two year"])

plt.legend()
plt.show()
```

```python
print(f'Month-to-Month: {round(5157 / 6785 * 100 , 2)}% Churn Rate')
print(f'One Year: {round(1044 / 3946 * 100 , 2)}% Churn Rate')
print(f'Two Year: {round(642 / 2426 * 100 , 2)}% Churn Rate')
```

# Contract Type VS Churn Rate

We are able to observe that contracts that are ran Month-To-Month has the highest Churn Rate. From this result, we are able to infer that customers tend to cancel more on month-to-month contracts compared to one-year or two-year contracts.

The cause of this may be due to many aspects. Perhaps customers just tend to forgets to cancel subscriptions contracts that are One Year or Two years..
